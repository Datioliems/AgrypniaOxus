```python
import tensorflow as tf
print('TF:', tf.__version__, '| GPU:', tf.config.list_physical_devices('GPU'))

CONFIG = {
    'image_size': 64,
    'batch_size': 64,
    'epochs': 25,
    'learning_rate': 1e-3,
    'dropout': 0.3,
    'split': {'train': 0.70, 'valid': 0.20, 'test': 0.10},
    'seed': 42,
    'closed_class_weight': 1.3,   # uu tien recall eyes_closed
    'aug': {'flip': 'horizontal', 'rotation': 0.05, 'zoom': 0.10, 'contrast': 0.20, 'brightness': 0.20},
}
CONFIG
```

```python
from google.colab import drive
drive.mount('/content/drive')

import zipfile, time, os
ZIP = '/content/drive/MyDrive/dataset (2).zip'   # sua neu de cho khac
t = time.time()
with zipfile.ZipFile(ZIP) as z:
    z.extractall('/content')
SRC = '/content/dataset'
print('Giai nen xong:', round(time.time()-t), 's ->', os.listdir(SRC))
```

```python
import os, re, random, shutil
from collections import defaultdict

CLASSES = ['eyes_closed', 'eyes_open']
OUT = '/content/dataset_subject'
random.seed(CONFIG['seed'])

# gom anh theo subject (parse s#### tu ten file MRL)
by_subject = defaultdict(list)   # subject -> [(class, path)]
for split in ['train', 'val', 'valid', 'test']:
    for c in CLASSES:
        d = f'{SRC}/{split}/{c}'
        if not os.path.isdir(d):
            continue
        for fn in os.listdir(d):
            m = re.match(r'(s\d+)', fn)
            subj = m.group(1) if m else fn   # fallback: coi file la 1 subject
            by_subject[subj].append((c, os.path.join(d, fn)))

subjects = list(by_subject.keys())
random.shuffle(subjects)
n = len(subjects)
n_tr = int(n * CONFIG['split']['train'])
n_va = int(n * CONFIG['split']['valid'])
split_subj = {'train': subjects[:n_tr], 'valid': subjects[n_tr:n_tr+n_va], 'test': subjects[n_tr+n_va:]}
print(f'Tong subject: {n} -> train {len(split_subj["train"])}, valid {len(split_subj["valid"])}, test {len(split_subj["test"])}')

if os.path.exists(OUT):
    shutil.rmtree(OUT)
counts = defaultdict(lambda: defaultdict(int))
for split, subs in split_subj.items():
    for subj in subs:
        for c, path in by_subject[subj]:
            dst = f'{OUT}/{split}/{c}'
            os.makedirs(dst, exist_ok=True)
            shutil.copy2(path, f'{dst}/{os.path.basename(path)}')
            counts[split][c] += 1
BASE = OUT
print('\nSo anh moi tap:')
for split in ['train', 'valid', 'test']:
    tot = sum(counts[split].values())
    print(f'  {split}: {dict(counts[split])} (tong {tot})')
```

```python
IMG, BATCH = CONFIG['image_size'], CONFIG['batch_size']
AUTOTUNE = tf.data.AUTOTUNE

def load(split, shuffle):
    return tf.keras.utils.image_dataset_from_directory(
        f'{BASE}/{split}', image_size=(IMG, IMG), batch_size=BATCH,
        label_mode='categorical', shuffle=shuffle, seed=CONFIG['seed'])

train_ds = load('train', True)
val_ds   = load('valid', False)
test_ds  = load('test', False)
class_names = train_ds.class_names
print('Classes:', class_names)   # ['eyes_closed', 'eyes_open']

norm = lambda x, y: (tf.cast(x, tf.float32) / 255.0, y)
train_ds = train_ds.map(norm, num_parallel_calls=AUTOTUNE)
val_ds   = val_ds.map(norm, num_parallel_calls=AUTOTUNE)
test_ds  = test_ds.map(norm, num_parallel_calls=AUTOTUNE)

a = CONFIG['aug']
aug = tf.keras.Sequential([
    tf.keras.layers.RandomFlip(a['flip']),
    tf.keras.layers.RandomRotation(a['rotation']),
    tf.keras.layers.RandomZoom(a['zoom']),
    tf.keras.layers.RandomContrast(a['contrast']),
    tf.keras.layers.RandomBrightness(a['brightness'], value_range=(0.0, 1.0)),
], name='augmentation')
train_ds = train_ds.map(lambda x, y: (aug(x, training=True), y), num_parallel_calls=AUTOTUNE).prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)
test_ds = test_ds.prefetch(AUTOTUNE)
```

```python
n_closed = counts['train']['eyes_closed']
n_open = counts['train']['eyes_open']
total = n_closed + n_open
class_weight = {
    0: (total / (2 * n_closed)) * CONFIG['closed_class_weight'],  # 0 = eyes_closed
    1: (total / (2 * n_open)),                                    # 1 = eyes_open
}
print('class_weight:', class_weight)
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Input((IMG, IMG, 3)),
    tf.keras.layers.Conv2D(32, 3, activation='relu'),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(64, 3, activation='relu'),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(128, 3, activation='relu'),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dropout(CONFIG['dropout']),
    tf.keras.layers.Dense(len(class_names), activation='softmax'),
])
model.compile(
    optimizer=tf.keras.optimizers.Adam(CONFIG['learning_rate']),
    loss='categorical_crossentropy',
    metrics=['accuracy',
             tf.keras.metrics.Precision(name='precision'),
             tf.keras.metrics.Recall(name='recall')],
)
model.summary()
```

```python
callbacks = [
    tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True, verbose=1),
    tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=2, min_lr=1e-5, verbose=1),
    tf.keras.callbacks.ModelCheckpoint('best_cnn.keras', monitor='val_recall', mode='max', save_best_only=True, verbose=1),
]
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=CONFIG['epochs'],
    class_weight=class_weight,
    callbacks=callbacks,
    verbose=1,
)
```

```python
import matplotlib.pyplot as plt
h = history.history
fig, ax = plt.subplots(1, 3, figsize=(16, 4))
ax[0].plot(h['accuracy'], label='train'); ax[0].plot(h['val_accuracy'], label='val'); ax[0].set_title('Accuracy'); ax[0].legend()
ax[1].plot(h['loss'], label='train'); ax[1].plot(h['val_loss'], label='val'); ax[1].set_title('Loss'); ax[1].legend()
ax[2].plot(h['recall'], label='train'); ax[2].plot(h['val_recall'], label='val'); ax[2].set_title('Recall'); ax[2].legend()
plt.show()
gap = h['accuracy'][-1] - h['val_accuracy'][-1]
print(f'Train-Val acc gap: {gap:.3f}  ({"OK, it overfit" if gap < 0.04 else "canh bao overfit"})')
```

```python
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

probs = model.predict(test_ds)
y_true = np.concatenate([y.numpy().argmax(1) for _, y in test_ds])
y_pred = probs.argmax(1)

print('=== Danh gia mac dinh (threshold 0.5) ===')
report = classification_report(y_true, y_pred, target_names=class_names, digits=4, output_dict=True)
print(classification_report(y_true, y_pred, target_names=class_names, digits=4))
cm = confusion_matrix(y_true, y_pred)
ConfusionMatrixDisplay(cm, display_labels=class_names).plot(); plt.show()

print('\n=== Threshold tuning cho eyes_closed (index 0) ===')
p_closed = probs[:, 0]
yt_closed = (y_true == 0).astype(int)
rows = []
for thr in [0.25, 0.30, 0.35, 0.40, 0.45, 0.50]:
    pred = (p_closed >= thr).astype(int)
    tp = int(((pred == 1) & (yt_closed == 1)).sum())
    fn = int(((pred == 0) & (yt_closed == 1)).sum())
    fp = int(((pred == 1) & (yt_closed == 0)).sum())
    rec = tp / (tp + fn + 1e-9); prec = tp / (tp + fp + 1e-9)
    rows.append((thr, round(rec, 4), round(prec, 4)))
    print(f'  thr={thr}: recall_closed={rec:.4f}  precision_closed={prec:.4f}')
cand = [r for r in rows if r[1] >= 0.98]
best_thr = (min(cand, key=lambda r: r[0])[0] if cand else 0.5)
print('=> Threshold deploy chon:', best_thr)
```

```python
import json, os
from datetime import datetime

run = {
    'run_time': datetime.now().strftime('%Y-%m-%d %H:%M:%S'),
    'config': CONFIG,
    'class_weight': {str(k): round(v, 3) for k, v in class_weight.items()},
    'split_counts': {s: dict(counts[s]) for s in ['train', 'valid', 'test']},
    'epochs_run': len(history.history['loss']),
    'final_train_acc': round(float(history.history['accuracy'][-1]), 4),
    'final_val_acc': round(float(history.history['val_accuracy'][-1]), 4),
    'test_accuracy': round(float(report['accuracy']), 4),
    'recall_eyes_closed': round(float(report['eyes_closed']['recall']), 4),
    'precision_eyes_closed': round(float(report['eyes_closed']['precision']), 4),
    'recall_eyes_open': round(float(report['eyes_open']['recall']), 4),
    'precision_eyes_open': round(float(report['eyes_open']['precision']), 4),
    'confusion_matrix': cm.tolist(),
    'threshold_tuning': [{'thr': r[0], 'recall_closed': r[1], 'precision_closed': r[2]} for r in rows],
    'deploy_threshold_closed': best_thr,
}

DRIVE = '/content/drive/MyDrive'
all_runs_path = f'{DRIVE}/cnn_eval_runs.json'
all_runs = json.load(open(all_runs_path)) if os.path.exists(all_runs_path) else []
all_runs.append(run)
json.dump(all_runs, open(all_runs_path, 'w'), indent=2, ensure_ascii=False)

md_path = f'{DRIVE}/cnn_eval_summary.md'
with open(md_path, 'a', encoding='utf-8') as f:
    f.write(f"\n## Run {run['run_time']}\n\n")
    f.write(f"- Epochs chay: {run['epochs_run']}/{CONFIG['epochs']}\n")
    f.write(f"- Test accuracy: **{run['test_accuracy']}**\n")
    f.write(f"- Recall eyes_closed: **{run['recall_eyes_closed']}** | precision: {run['precision_eyes_closed']}\n")
    f.write(f"- Recall eyes_open: {run['recall_eyes_open']} | precision: {run['precision_eyes_open']}\n")
    f.write(f"- Train-Val acc: {run['final_train_acc']} / {run['final_val_acc']}\n")
    f.write(f"- Threshold deploy (closed): {run['deploy_threshold_closed']}\n")
    f.write(f"- Confusion matrix: {run['confusion_matrix']}\n")

print('Da ghi bao cao:', all_runs_path, f'({len(all_runs)} lan chay) +', md_path)
from google.colab import files
files.download(md_path)
files.download(all_runs_path)
```
