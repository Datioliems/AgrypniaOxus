# Sinh DANH MỤC TỪ VIẾT TẮT
Xuất file `outputs/DanhMuc_TuVietTat.docx` — đúng format IS54A (TNR 13pt, căn đều, bảng 3 cột).

```python
# -*- coding: utf-8 -*-
import os
import docx
from docx.shared import Pt, Cm
from docx.enum.text import WD_ALIGN_PARAGRAPH
from docx.enum.table import WD_TABLE_ALIGNMENT

os.makedirs('outputs', exist_ok=True)

# ── Danh sách từ viết tắt — sắp xếp theo bảng chữ cái (A-Z) ─────────────────
ABBREVS = [
    # (Từ viết tắt, Tiếng Anh đầy đủ, Nghĩa tiếng Việt)
    ('AI',      'Artificial Intelligence',
                'Trí tuệ nhân tạo'),
    ('AP',      'Average Precision',
                'Độ chính xác trung bình'),
    ('API',     'Application Programming Interface',
                'Giao diện lập trình ứng dụng'),
    ('APK',     'Android Package Kit',
                'Gói cài đặt ứng dụng Android'),
    ('CNN',     'Convolutional Neural Network',
                'Mạng nơ-ron tích chập'),
    ('CPU',     'Central Processing Unit',
                'Đơn vị xử lý trung tâm'),
    ('DETR',    'Detection Transformer',
                'Mô hình phát hiện đối tượng dựa trên Transformer'),
    ('EAR',     'Eye Aspect Ratio',
                'Tỉ lệ khung mắt'),
    ('FN',      'False Negative',
                'Âm tính giả (bỏ sót)'),
    ('FP',      'False Positive',
                'Dương tính giả (báo nhầm)'),
    ('FPS',     'Frames Per Second',
                'Số khung hình mỗi giây'),
    ('GPS',     'Global Positioning System',
                'Hệ thống định vị toàn cầu'),
    ('GPU',     'Graphics Processing Unit',
                'Đơn vị xử lý đồ hoạ'),
    ('IoU',     'Intersection over Union',
                'Tỉ lệ giao trên hợp'),
    ('mAP',     'mean Average Precision',
                'Độ chính xác trung bình tổng hợp'),
    ('MAR',     'Mouth Aspect Ratio',
                'Tỉ lệ khung miệng'),
    ('ML',      'Machine Learning',
                'Học máy'),
    ('MRL',     'Machine Reading Lab',
                'Tên bộ dữ liệu trạng thái mắt (MRL Eye Dataset)'),
    ('RF-DETR', 'Receptive Field Detection Transformer',
                'Mô hình DETR dựa trên trường tiếp nhận'),
    ('ROI',     'Region of Interest',
                'Vùng quan tâm'),
    ('RT-DETR', 'Real-Time Detection Transformer',
                'Mô hình phát hiện Transformer thời gian thực'),
    ('SDK',     'Software Development Kit',
                'Bộ công cụ phát triển phần mềm'),
    ('SMS',     'Short Message Service',
                'Dịch vụ tin nhắn ngắn'),
    ('TFLite',  'TensorFlow Lite',
                'TensorFlow phiên bản nhúng (trên thiết bị di động)'),
    ('TN',      'True Negative',
                'Âm tính thật'),
    ('TP',      'True Positive',
                'Dương tính thật'),
    ('UI',      'User Interface',
                'Giao diện người dùng'),
    ('YOLO',    'You Only Look Once',
                'Thuật toán phát hiện đối tượng một lần nhìn'),
]

# ── Tạo document ─────────────────────────────────────────────────────────────
doc = docx.Document()
sty = doc.styles['Normal']
sty.font.name = 'Times New Roman'
sty.font.size = Pt(13)

# Tiêu đề
title = doc.add_paragraph()
title.alignment = WD_ALIGN_PARAGRAPH.CENTER
tr = title.add_run('DANH MỤC TỪ VIẾT TẮT')
tr.bold = True
tr.font.size = Pt(14)

doc.add_paragraph()  # khoảng trống

# Header bảng
COLS = ['STT', 'Từ viết tắt', 'Tiếng Anh đầy đủ', 'Nghĩa tiếng Việt']
WIDTHS = [Cm(1.2), Cm(2.4), Cm(6.8), Cm(8.1)]  # tổng ~18.5cm

tbl = doc.add_table(rows=1, cols=4)
tbl.style = 'Table Grid'
tbl.alignment = WD_TABLE_ALIGNMENT.CENTER

hdr = tbl.rows[0].cells
for i, (cell, col, w) in enumerate(zip(hdr, COLS, WIDTHS)):
    cell.width = w
    p = cell.paragraphs[0]
    p.alignment = WD_ALIGN_PARAGRAPH.CENTER
    run = p.add_run(col)
    run.bold = True
    run.font.size = Pt(12)

# Dữ liệu
for idx, (abbr, en, vi) in enumerate(ABBREVS, start=1):
    row = tbl.add_row().cells
    for i, w in enumerate(WIDTHS):
        row[i].width = w

    # STT
    p = row[0].paragraphs[0]
    p.alignment = WD_ALIGN_PARAGRAPH.CENTER
    p.add_run(str(idx)).font.size = Pt(12)

    # Từ viết tắt (đậm, căn giữa)
    p = row[1].paragraphs[0]
    p.alignment = WD_ALIGN_PARAGRAPH.CENTER
    r2 = p.add_run(abbr)
    r2.bold = True
    r2.font.size = Pt(12)

    # Tiếng Anh
    row[2].paragraphs[0].add_run(en).font.size = Pt(12)

    # Nghĩa tiếng Việt
    row[3].paragraphs[0].add_run(vi).font.size = Pt(12)

doc.save('outputs/DanhMuc_TuVietTat.docx')
print(f'OK — {len(ABBREVS)} từ viết tắt -> outputs/DanhMuc_TuVietTat.docx')
```
